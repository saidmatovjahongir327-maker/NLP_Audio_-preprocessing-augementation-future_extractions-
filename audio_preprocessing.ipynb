```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import IPython.display as ipd  #  to play sound in the nootbok
import librosa
import librosa.display
```

```python
# list of all audio files
import os
audio_files = os.listdir("audios")
audio_files
```

```python
data , sr = librosa.load("audios/JANAGA, TENCA - Не из вашего круга.mp3")
```

```python
data
```

```python
sr
```

```python
print(f'y: {data[:10]}')
print(f'shape y: {data.shape}')
print(f'sr: {sr}')
```

```python
pd.Series(data).plot(figsize=(10,5),
                            lw=1,
                            title='Raw Audio Example',
                            )
plt.show()
```

```python
melspec = librosa.feature.melspectrogram(y=data, sr=sr)
```

```python
log_s = librosa.amplitude_to_db(melspec)
```

```python
plt.figure(figsize=(12,4))
librosa.display.specshow(log_s,sr=sr, x_axis="time", y_axis="mel")
```
