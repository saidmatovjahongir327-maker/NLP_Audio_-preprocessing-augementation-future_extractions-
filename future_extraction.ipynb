```python
def extract_standart_devitation(y,sr=22050):
    return librosa.feature.spectral_centroid(y=y,sr=sr).std()
```

```python
def extract_median(y,sr=22050):
    return librosa.feature.spectral_centroid(y=y,sr=sr)[0]
```

```python
def extract_kurtosis(y):
    return kurtosis(y)
```

```python
def extract_spectral_fltness(y):
    return librosa.feature.spectral_flatness(y=y)[0]
def extract_mode(y):
    return mode(y)[0]

def extract_centroid(y,sr):
    return librosa.feature.spectral_centroid(y=y,sr=sr)[0]
def extract_peak(y):
    return librosa.feature.spectral_centroid(y=y)[0]
def extract_mean_function(y,sr):
    return librosa.feature.spectral_centroid(y=y,sr=sr).mean()
def extract_minimum(y,sr):
    return librosa.feature.spectral_centroid(y=y,sr=sr).min()
def extract_maximum(y,sr):
    return librosa.feature.spectral_centroid(y=y,sr=sr).max()
def extract_mean_dom(y,sr):
    return librosa.feature.spectral_bandwidth(y=y,sr=sr).mean()
def extract_minimum_dom(y,sr):
    return librosa.feature.spectral_bandwidth(y=y,sr=sr).min()  
def extract_maximum_dom(y,sr):
    return librosa.feature.spectral_bandwidth(y=y,sr=sr).max()
def extract_dfrange(maxdom, mindom):
    return maxdom - mindom  
def extract_modindex(y,sr):
    return librosa.feature.tempogram(y=y,sr=sr).mean()
```

```python
def extract_features(audio_file):
    y, sr = librosa.load(audio_file)
    
    features = {
        "mean_frequency": extract_mean_frequency(y, sr),
        "standard_deviation": extract_standart_devitation(y, sr),
        "median": extract_median(y, sr),
        "kurtosis": extract_kurtosis(y),
        "spectral_flatness": extract_spectral_fltness(y),
        "mode": extract_mode(y),
        "centroid": extract_centroid(y, sr),
        "peak": extract_peak(y),
        "mean_function": extract_mean_function(y, sr),
        "minimum": extract_minimum(y, sr),
        "maximum": extract_maximum(y, sr),
        "mean_dom": extract_mean_dom(y, sr),
        "minimum_dom": extract_minimum_dom(y, sr),
        "maximum_dom": extract_maximum_dom(y, sr),
        "dfrange": extract_dfrange(extract_maximum_dom(y, sr), extract_minimum_dom(y, sr)),
        "modindex": extract_modindex(y, sr)
    }
    
    return features
```

```python
def process_audio_directory(input_dir, output_file):
    # Initalize an empty list to store the extracted features
    features_list = []

    #Loop through all the audio files in the input directory
    for filename in os.listdir(input_dir):
       if filename.endswith(".mp3") or filename.endswith(".wav"):
           audio_file = os.path.join(input_dir, filename)
           features = extract_features(audio_file)
           features_list.append(features)
    # Convert the list of features to a DataFrame
    features_df = pd.DataFrame(features_list)
    # Save the DataFrame to a CSV file
    features_df.to_csv(output_file, index=False)

#Example usage
input_directory = "audios"
output_csv_file = "extracted_feature1.csv"
process_audio_directory(input_directory, output_csv_file)
```
